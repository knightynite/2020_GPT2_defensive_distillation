## 2. Gradient norms (the smoking gun)

Distillation with T=100 shrinks input-space gradients by ~1/T (≈100×). FGSM uses sign-of-gradient → noisy at small magnitudes → apparent defense. Optimization-based attacks (CW with logit-difference loss) bypass this entirely because they don't operate on the cross-entropy gradient.
